```python
def shortlist(candidates, top_n=15):
    """فیلتر درشت: نمادهای مثبت امروز، مرتب بر اساس حجم."""
    positive = []
    for c in candidates:
        if c["yday"] > 0 and c["last"] > c["yday"]:
            positive.append(c)

    positive.sort(key=lambda c: c["volume"], reverse=True)
    return positive[:top_n]
```
